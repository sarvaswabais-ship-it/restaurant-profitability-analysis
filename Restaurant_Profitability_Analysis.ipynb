```python
completed_order_items = order_items[
    order_items["order_id"].isin(completed_orders["order_id"])
]

completed_order_items["product_id"].nunique()
```

```python
product_summary = completed_order_items.merge(
    product[
        ["product_id", "product_name", "category", "unit_food_cost"]
    ],
    on="product_id",
    how="left"
)

product_summary["revenue"] = (
    product_summary["quantity"] *
    product_summary["unit_price"]
)

product_summary["food_cost"] = (
    product_summary["quantity"] *
    product_summary["unit_food_cost"]
)

product_summary["profit"] = (
    product_summary["revenue"] -
    product_summary["food_cost"]
)

product_summary = product_summary.groupby(
    ["product_id", "product_name", "category"],
    as_index=False
).agg(
    quantity=("quantity", "sum"),
    revenue=("revenue", "sum"),
    food_cost=("food_cost", "sum"),
    profit=("profit", "sum")
)

product_summary["profit_margin_pct"] = (
    product_summary["profit"] /
    product_summary["revenue"] * 100
)
```

```python
total_revenue = product_summary["revenue"].sum()
total_profit = product_summary["profit"].sum()

total_revenue, total_profit
```

```python
product_summary["revenue_contribution_pct"] = (
    product_summary["revenue"] /
    total_revenue * 100
)

product_summary["profit_contribution_pct"] = (
    product_summary["profit"] /
    total_profit * 100
)
```

```python
product_summary[
    [
        "product_id",
        "product_name",
        "category",
        "revenue",
        "profit",
        "profit_margin_pct",
        "revenue_contribution_pct",
        "profit_contribution_pct"
    ]
].sort_values(
    "profit_contribution_pct",
    ascending=False
).head()
```

```python
category_summary = product_summary.groupby(
    "category",
    as_index=False
)[
    ["revenue", "profit"]
].sum()
```
